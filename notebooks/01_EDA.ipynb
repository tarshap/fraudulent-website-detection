# 01. Exploratory Data Analysis (EDA)
## Fraudulent Website Detection System using XGBoost

This notebook explores the **PhiUSIIL Phishing URL Dataset** (UCI ML Repository ID 967) to inspect feature distributions, correlations, and class balance.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
%matplotlib inline
```

### 1. Load Dataset

```python
data_path = os.path.join("..", "data", "raw", "phiusiil.csv")
if os.path.exists(data_path):
    df = pd.read_csv(data_path)
    print(f"Dataset loaded successfully: {df.shape[0]:,} rows, {df.shape[1]} columns.")
    display(df.head())
else:
    print(f"Dataset not found at {data_path}. Run 'python src/download_dataset.py' to download.")
```

### 2. Target Variable Distribution

```python
target_candidates = [c for c in df.columns if c.lower() in ['label', 'target', 'class', 'phishing', 'result']]
if target_candidates:
    target_col = target_candidates[0]
    print(f"Target column: {target_col}")
    print(df[target_col].value_counts())
    
    plt.figure(figsize=(6, 4))
    sns.countplot(data=df, x=target_col, palette="Set2")
    plt.title("Class Distribution (Legitimate vs Phishing)")
    plt.xlabel("Class Label")
    plt.ylabel("Count")
    plt.show()
else:
    print("Could not detect target column automatically.")
```

### 3. Check Missing Values and Duplicates

```python
print(f"Duplicate rows: {df.duplicated().sum():,}")
print(f"Total missing values: {df.isnull().sum().sum():,}")
```

### 4. Correlation Analysis

```python
numeric_df = df.select_dtypes(include=[np.number])
if not numeric_df.empty:
    plt.figure(figsize=(12, 10))
    sample_corr = numeric_df.iloc[:, :15].corr()
    sns.heatmap(sample_corr, annot=True, cmap="coolwarm", fmt=".2f", cbar=True)
    plt.title("Correlation Heatmap (Sample Numerical Features)")
    plt.show()
```
